```python
# --------------------------------------------------
# 1. IMPORTAR LAS BIBLIOTECAS
# --------------------------------------------------

# Importar NumPy para generar valores y realizar cálculos numéricos.
import numpy as np

# Importar Matplotlib para representar la curva y la recta tangente.
import matplotlib.pyplot as plt

# --------------------------------------------------
# 2. DEFINIR LA CURVA IMPLÍCITA
# --------------------------------------------------

# La ecuación estudiada es y**2 = x**2*(5 - x**2).

# La curva tiene valores reales cuando -sqrt(5) <= x <= sqrt(5).
# np.linspace genera 500 valores distribuidos uniformemente.
x = np.linspace(-np.sqrt(5), np.sqrt(5), 500)

# Calcular el contenido de la raíz cuadrada:
# valor = x**2*(5 - x**2).
valor = x**2 * (5 - x**2)

# CORRECCIÓN: Evitar que errores de precisión de punto flotante
# dejen valores negativos infinitesimales al borde del dominio.
valor = np.clip(valor, 0, None)

# Calcular la rama superior de la curva mediante la raíz positiva.
y_superior = np.sqrt(valor)

# Calcular la rama inferior de la curva mediante la raíz negativa.
y_inferior = -np.sqrt(valor)

# --------------------------------------------------
# 3. DEFINIR EL PUNTO DE TANGENCIA
# --------------------------------------------------

# Coordenada horizontal del punto P = (2, 2).
xp = 2

# Coordenada vertical del punto P = (2, 2).
yp = 2

# Pendiente calculada mediante derivación implícita:
# dy/dx = (10*x - 4*x**3)/(2*y).
# Al evaluar en x = 2 e y = 2, obtenemos m = -3.
m = -3

# --------------------------------------------------
# 4. CALCULAR LA RECTA TANGENTE
# --------------------------------------------------

# Generar 200 valores de x cerca del punto de tangencia.
x_recta = np.linspace(1.0, 2.3, 200) cambio de valores ojo

# Aplicar la ecuación punto-pendiente:
# y - yp = m*(x - xp).
# Despejando: y = m*(x - xp) + yp.
y_recta = m * (x_recta - xp) + yp

# --------------------------------------------------
# 5. REPRESENTAR GRÁFICAMENTE LA CURVA
# --------------------------------------------------

# Crear una figura de 10 pulgadas de ancho y 7 de alto.
plt.figure(figsize=(10, 7))

# Dibujar la rama superior de la curva.
plt.plot(x, y_superior, label="Rama superior de la curva", color="blue")

# Dibujar la rama inferior de la curva.
plt.plot(x, y_inferior, label="Rama inferior de la curva", color="blue")

# --------------------------------------------------
# 6. REPRESENTAR LA RECTA TANGENTE
# --------------------------------------------------

# Dibujar la recta tangente con línea discontinua.
plt.plot(x_recta, y_recta, label="Recta tangente: y = -3x + 8", color="red", linestyle="--")

# --------------------------------------------------
# 7. MARCAR EL PUNTO P
# --------------------------------------------------

# Dibujar el punto de tangencia P = (2, 2).
plt.scatter(xp, yp, label="Punto P = (2, 2)", color="black", zorder=5)

# Añadir una etiqueta para identificar el punto en la gráfica.
plt.annotate("P = (2, 2)", (xp, yp), xytext=(10, 10), textcoords="offset points")

# --------------------------------------------------
# 8. CONFIGURAR LOS EJES Y LA PRESENTACIÓN
# --------------------------------------------------

# Dibujar el eje horizontal y = 0.
plt.axhline(0, color="gray", linestyle=":")

# Dibujar el eje vertical x = 0.
plt.axvline(0, color="gray", linestyle=":")

# Añadir el título de la gráfica.
plt.title("Curva implícita y recta tangente")

# Identificar el eje horizontal.
plt.xlabel("x: consumo de energía (MWh)")

# Identificar el eje vertical.
plt.ylabel("y: producción diaria (miles de unidades)")

# Mostrar una cuadrícula para facilitar la lectura.
plt.grid(True)

# Mostrar la leyenda que identifica las líneas y el punto.
plt.legend()

# Ajustar los límites horizontales de la ventana gráfica.
plt.xlim(-2.5, 2.5) cambio de valores ojo

# Ajustar los límites verticales de la ventana gráfica.
plt.ylim(-3, 4) cambie valores ojo

# Mostrar la gráfica resultante.
plt.show()
```
