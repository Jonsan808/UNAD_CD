```python

# 1. IMPORTAR LAS LIBRERÍAS

# NumPy permite crear conjuntos de valores numéricos que utilizaremos para representar el eje x.
import numpy as np

# Matplotlib permite crear y mostrar la gráfica.
import matplotlib.pyplot as plt

# 2. CREAR LOS VALORES DE x PARA CADA TRAMO

# Primera rama:
# f(x) = 2x + 4
# Condición: x < 1
# Creamos valores desde -2 hasta 1.
# endpoint=False evita incluir exactamente el valor x=1.
x1 = np.linspace(-2, 1, 300, endpoint=False)

# Segunda rama:
# f(x) = x**2 + 2
# Condición: 1 ≤ x < 3
# Comenzamos en x=1 porque este punto sí pertenece a la segunda rama.
# endpoint=False evita incluir x=3 porque la condición indica que x debe ser menor que 3.
x2 = np.linspace(1, 3, 300, endpoint=False)

# Tercera rama:
# f(x) = x + 8
# Condición: x ≥ 3
# Comenzamos en x=3 porque este punto sí pertenece a la tercera rama.
x3 = np.linspace(3, 6, 300)

# 3. CALCULAR LOS VALORES DE (y)

# Calculamos la primera rama de la función:
# y = 2x + 4
y1 = 2 * x1 + 4

# Calculamos la segunda rama:
# y = x**2 + 2
y2 = x2**2 + 2

# Calculamos la tercera rama:
# y = x + 8
y3 = x3 + 8

# 4. CREAR LA GRAFICA

# Creamos una figura de 10 pulgadas de ancho por 6 pulgadas de alto.
plt.figure(figsize=(10, 6))

# 5. GRAFICAR LA PRIMERA RAMA

# Graficamos:
# f(x) = 2x + 4, cuando x < 1
plt.plot(x1, y1, label="2x + 4, si x < 1", color="red")

# 6. GRAFICAR LA SEGUNDA RAMA

# Graficamos:
# f(x) = x**2 + 2, cuando 1 ≤ x < 3
plt.plot(x2, y2, label="x² + 2, si 1 ≤ x < 3", color="blue")

# 7. GRAFICAR LA TERCERA RAMA

# Graficamos:
# f(x) = x + 8, cuando x ≥ 3
plt.plot(x3, y3, label="x + 8, si x ≥ 3", color="green")

# 8. MARCAR EL PUNTO ABIERTO EN x = 1

# Cuando x se acerca a 1 por la izquierda:
# 2(1) + 4 = 6
# Por lo tanto, el punto es (1, 6).
# Es un punto abierto porque la primera rama tiene:
# x < 1
plt.scatter(1, 6, facecolors="white", edgecolors="black", s=100, zorder=5, label="Punto abierto (1, 6)")

# 9. MARCAR EL PUNTO CERRADO EN x = 1

# En x = 1 utilizamos la segunda rama:
# 1**2 + 2 = 3
# Por lo tanto, el punto es (1, 3).
# Es un punto cerrado porque:
# 1 ≤ x < 3
plt.scatter(1, 3, color="black", s=80, zorder=6, label="Punto cerrado (1, 3)")

# 10. MARCAR EL PUNTO ABIERTO EN x = 3

# La segundad rama se aproxima a:
# 3**2 + 2 = 11
# El punto (3, 11) es abierto para esta rama
# porque la condición es x < 3.
plt.scatter( 3, 11, facecolors="white", edgecolors="black", s=100, zorder=5, label="Punto abierto (3, 11)")

# 11. MARCAR EL PUNTO CERRADO EN x = 3

# La tercera rama comienza en:
# 3 + 8 = 11
# Por lo tanto:
# f(3) = 11
# Este punto es cerrado porque la condición es:
# x ≥ 3
plt.scatter( 3, 11, color="black", s=50, zorder=6, label="Punto cerrado (3, 11)")

# Se dibuja el eje de simetría de la parábola en x=1 con color morado y con líneas y puntos
plt.axvline(1, linestyle="-.", label="eje de simetría x=1", color="purple",)

# Se dibuja el eje de simetría de la parábola en x=3 con color azul claro y con líneas y puntos
plt.axvline(3, linestyle="-.", label="eje de simetría x=3", color="cyan",)

# 12. AGREGAR EL EJE HORIZONTAL

# Esta línea permite visualizar el eje x con color negro y con dotted línea punteada.
plt.axhline(0, color="black", linestyle="dotted")

# 13. AGREGAR EL EJE VERTICAL

# Esta línea permite visualizar el eje y con color negro y con dotted línea punteada.
plt.axvline(0, color="black", linestyle="dotted")

# 14. COLOCAR EL TÍTULO

# Agregamos un título que identifica la gráfica.
plt.title("Continuidad de una función a trozos")

# 15. NOMBRAR EL EJE x
plt.xlabel("x")

# 16. NOMBRAR EL EJE y
plt.ylabel("f(x)")

# 17. ESTABLECER EL INTERVALO DEL EJE x
# La gráfica mostrará valores desde -2 hasta 6.
plt.xlim(-2, 6)

# 18. ESTABLECER EL INTERVALO DEL EJE y

# Establecemos un rango vertical adecuado
# La gráfica mostrará valores desde -1 hasta 15 para visualizar los tres tramos.
plt.ylim(-1, 15)

# 19. ACTIVAR LA CUADRÍCULA

# La cuadrícula facilita la lectura de los puntos
# y valores de la gráfica.
plt.grid(True)

# 20. MOSTRAR LA LEYENDA

# La leyenda identifica cada tramo y los puntos
# abiertos y cerrados.
plt.legend()

# 21. MOSTRAR LA GRÁFICA

# Finalmente, mostramos la gráfica en Google Colab.
plt.show()
```
